```python
#Importando Bibliotecas
import pandas as pd
import numpy as np
```

```python
#Carregando Arquivo csv
df = pd.read_csv("/content/Base Varejo.csv",sep=';',encoding='latin-1')
```

```python
#Mostrar informações iniciais
print("--- Informações da Base ---")
print("Prinmeiras 10 linhas do dataset:")
df.head(10)
```

```python
#Verificando os tipos de dados nas colunas
print(f"Total de registros: {df.shape[0]}")
print(f"Total de colunas: {df.shape[1]}")
print(" Tipo de dado de cada coluna:")
print(df.dtypes)
```

```python
# Verificação de nulos e duplicatas
print("\n--- Verificação de Problemas ---")
print("Valores nulos por coluna:\n", df.isnull().sum())
print("Total de duplicatas:", df.duplicated().sum())
```

Após analise inicial dos dados podemos observar 4 colunas com valores Nulos superior a 90% dos registros da referida coluna. Essas colunas deverão ser removidas na etapa de limpeza. A coluna DATA contem dados do tipo object, esta coluna deve ser convertida para o tipo datetime para permitir análises com base no tempo. Com relação aos dados duplicados, foram encontrados 96.553 registros duplicados dos 830.000, esses dados deverão ser avaliados durante a etapa de limpeza para não haver distorção nas análises.

```python
#Convertendo coluna data
df['DATA'] = pd.to_datetime(df['DATA'], dayfirst=True, errors='coerce')
df['DATA'].info()
```

```python
#Removendo colunas com valores nulos
print("Dimensão do DF antes da remoção das colunas vazias:")
print(df.shape)
df = df.dropna(axis=1, how='all')
print("Dimensão do DF depois da remoção das colunas vazias:")
print(df.shape)
```

```python
#Removendo linhas duplicadas
print(f"Dimensão do DF antes da remoção das linhas duplicadas: {(df.shape)}")

df = df.drop_duplicates()
print(f"Dimensão do DF depois da remoção das linhas duplicadas: {(df.shape)}")
```

```python
#Tratando dados na coluna categoria
df['PR_CAT'] = df['PR_CAT'].fillna("SEM CATEGORIA")
df['PR_CAT'] = df['PR_CAT'].replace('#N/D', 'SEM CATEGORIA')
```

```python
#Salvando base de dados limpa
df.to_csv('/content/base_varejo_limpa.csv', index=False, sep=';')
```

Após a etapa de limpeza pode ser iniciada a analise estatística das colunas.

```python
# --- Estatística Descritiva ---
#Coluna de número de filhos
df['CL_FHL'].describe()
```

Podemos verificar que a média do número de filhos é 1 e o comprador com mais filhos é 4.

```python
# --- Agrupamentos ---
#Gênero com mais compras
df['CL_GENERO'].value_counts()
```

Se analisarmos o gênero do comprador podemos verificar que o sexo feminino efetuou o maior número das compras do banco de dados.

```python
#Categoria com mais pedidos
df.groupby('PR_CAT').size().sort_values(ascending=False)
```

Podemos verificar que os itens com maior volume de venda são Alimentos, Higiene e Limpeza.

```python
#Agrupando por gênero do comprador e categoria do produto
df.groupby(['CL_GENERO', 'PR_CAT']).size()
```

Ao fazer o agrupamento por gênero do comprador e categoria do produto podemos observar que o gênero feminino comprou mais que o masculino em todas as categorias.
